# ***Detecting Anomalous Driver Behavior in NYC Taxi Data***

# **1. Introduction**

**1.1. Project Overview**

This project focuses on the analysis of a large-scale dataset containing taxi ride records from New York City, collected throughout 2013. Each record represents a completed trip and includes detailed information such as timestamps, coordinates, trip distance, duration, payment method, and monetary values such as fare, tips, and total amount paid.

**1.2. Selected Research Question**

This project investigates the following research question:

*“Do any taxi drivers exhibit anomalous behavior compared to the overall driver population?”*

The goal is to identify drivers whose operational patterns significantly differ from the majority. In particular, we aim to detect anomalies in:

*   **Trip duration behavior:** drivers who consistently complete trips much faster or slower than expected for similar distances.
*   **Tipping behavior:** drivers associated with unusually high or unusually low tip amounts.
*   **Route patterns:** irregular or unexpected travel paths between similar pickup and drop-off areas.
*   **Potential fraud indicators:** combinations of abnormal duration, pricing, or routing that may suggest incorrect reporting or fare manipulation.

By analyzing these dimensions, we aim to highlight outlier drivers and explore possible explanations for their deviations, whether due to operational efficiency, passenger behavior, data errors, or potential fraud.





# **2. Dataset Description**

**2.1 Dataset Origin**

The dataset analyzed in this project originates from a large collection of New York City taxi ride records collected during the year 2013. The data was made publicly available as part of the DEBS 2015 Grand Challenge context and reflects real-world taxi operations across the NYC metropolitan area.

For this project, we use the 1% sample, which contains approximately 1.7 million taxi trips. This subset preserves the statistical properties, temporal patterns, and noise of the full dataset while enabling efficient experimentation within the constraints of the Colab environment.

**2.2. Schema Overview**

The dataset is provided as a compressed CSV file (taxi_rides_1pc.csv.gz). After loading the data using Spark, the schema is automatically inferred. The most relevant fields are:


*   **medallion:** Hashed identifier of the taxi vehicle
*   **hack_license:** Hashed identifier of the taxi driver
*   **pickup_datetime / dropoff_datetime:** Timestamps marking the start and end of the trip
*   **trip_time_in_secs:** Duration of the ride
*   **trip_distance:** Distance traveled in miles
*   **pickup_longitude / pickup_latitude:** GPS coordinates for pickup
*   **dropoff_longitude / dropoff_latitude:** GPS coordinates for drop-off
*   **payment_type:** Payment method (cash or credit)
*   **fare_amount, tip_amount, total_amount, surcharge, mta_tax, tolls_amount:** Monetary information about the trip










**2.3. Attribute Description and Relevance**

Below is a concise summary of the most relevant attributes for our research question:

**Driver-related fields**
*   **hack_license:** Used to group trips by driver and analyze behavioral deviations.

**Trip performance fields**
*   **trip_time_in_secs** and **trip_distance:**
Key features for detecting drivers who consistently take unusually long or short times for similar distances.

**Monetary fields**
*   **tip_amount:**
Essential for identifying drivers associated with unusually high or unusually low tipping patterns.
*   **fare_amount** and **total_amount:**
Useful for cross-validating pricing anomalies and potential signs of fraud.

**Spatial fields**
*   **pickup/dropoff latitude and longitude:**
Allow the computation of route irregularities or deviations from common travel patterns between similar locations.

**2.4. Data Quality and Limitations**

The dataset reflects real-world operational noise and includes several limitations that must be considered during analysis:

*   **Missing or invalid coordinates:**
Some trips have 0.0, extremely large, or otherwise unrealistic coordinates, which may distort spatial analysis.

*   **Incorrect monetary values:**
Fares, tips, or total amounts may contain errors introduced during manual entry by taxi drivers.

*   **Potential duplicate or malformed rides:**
Some events may be incomplete or inconsistent due to logging issues.

*   **Outliers in trip time or distance:**
Extremely short or long durations may represent erroneous entries rather than real trips.

*   **Reporting timing:**
All events are logged only at drop-off, not at pickup, which affects temporal alignment when analyzing driver behavior.

Given these constraints, data cleaning and filtering steps are essential before conducting anomaly detection.
